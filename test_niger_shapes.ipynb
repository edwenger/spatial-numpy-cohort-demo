```python
import os

from bokeh.models import GeoJSONDataSource
from bokeh.plotting import figure, show, output_notebook
import geopandas as gpd
```

```python

path = os.path.join("GRID3", "ner_adm_ignn_20230720_em_shp", "NER_admbnda_adm2_IGNN_20230720_em.shp")
lgas = gpd.read_file(path)
lgas["geometry"] = lgas["geometry"].to_crs(crs="EPSG:4326")

lgas = lgas[lgas.ADM1_FR.isin([
    "Zinder",
    "Maradi", 
    ])]
```

```python
lgas.iloc[0]
```

```python
output_notebook()

geo_source = GeoJSONDataSource(geojson=lgas.to_json())

p = figure(width=800, tooltips=[('LGA', '@ADM2_FR'), ('State', '@ADM1_FR')])
p.patches('xs', 'ys', source=geo_source, fill_alpha=0.4, fill_color="gray", line_color="black", line_width=0.5)

show(p)
```
